### **5. 정비 견적(만원)(REPAIR_EST) 컬럼의 이상치는 삭제하지 않고 정상 범위의 경계값으로 대체할 계획입니다.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, clip, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 아래 가이드에 따라 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df_raw
    - IQR 기준(K=1.5)으로 lower_fence, upper_fence 를 산출하세요
    - REPAIR_EST 값이 두 fence 를 벗어나면 가장 가까운 fence 값으로 바꾸세요(행은 삭제하지 않음)
    - 식별자 컬럼 CASE_ID 를 삭제한 사본에 적용해 df_clean 에 보관하세요
- **[정답05]** 처리 후 REPAIR_EST 의 **최댓값**을 소수점 아래 둘째 자리까지 작성하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요
- 배점 : 6점

---

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)
# 1. IQR 계산
q1 = df_raw['REPAIR_EST'].quantile(0.25)
q3 = df_raw['REPAIR_EST'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

df_clean = df_raw.drop(columns=['CASE_ID']).copy()
df_clean['REPAIR_EST'] = df_clean['REPAIR_EST'].clip(lower=lower_fence, upper=upper_fence)

# [정답05] `__`·`____` 자리를 채워 실행한 결과를 입력하세요
df_clean['REPAIR_EST'].max()
round(df_clean['REPAIR_EST'].max(),2)
````

````python
# (5) 정답을 입력하세요 — `정답05` 에 값만 넣고 실행하세요

정답05 = 310.625
````
