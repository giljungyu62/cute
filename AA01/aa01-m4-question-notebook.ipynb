### **5. 그래프로 확인한 결과, 경유지 수(N_STOPS) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 아래 안내대로 이상치를 처리하세요.**

- 앞 문항 결과 사용 : 2번 `df_raw`
- 대상 데이터프레임 : df_raw
- 경유지 수(N_STOPS)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=3.0)
- AI모델링에 불필요한 표본 식별번호(SAMPLE_ID) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 df_clean 라는 변수명으로 담으세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요 (글자는 따옴표로 감싸기)
- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = df_raw['N_STOPS'].quantile(0.25)
q3 = df_raw['N_STOPS'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 3.0 * iqr
upper_fence = q3 + 3.0 * iqr

df_clean = df_raw.drop( df_raw[(df_raw['N_STOPS'] > upper_fence) | (df_raw['N_STOPS'] < lower_fence)].index, axis=0)
df_clean = df_clean.drop(columns=['SAMPLE_ID'])
df_clean.reset_index(drop=True, inplace=True)
````

````python
# (5) 정답을 입력하세요 — `정답05` 에 값만 넣고 실행하세요

정답05 = 'reset_index'
````

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

model = Sequential([
    Dense(32, activation='relu', input_dim=X_train.shape[1]),
    Dropout(0.3),
    Dense(16, activation='relu'),
    Dense(1)
])

estop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
cb_list = [estop]

model.compile(optimizer='nadam', loss='mse', metrics=['mae', 'mse'])

history = model.fit(X_train, y_train,
                    epochs=30, batch_size=32,
                    validation_data=(X_valid, y_valid),
                    callbacks=cb_list)
````
