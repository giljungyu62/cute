### **5. 그래프로 확인한 결과, 가입 경과일(일)(Tenure_Days) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 다음 지침에 따라 이상치를 처리하세요.**

- 앞 문항 결과 사용 : 2번 `df`
- 대상 데이터프레임 : df
- 가입 경과일(일)(Tenure_Days)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=2.0)
- AI모델링에 불필요한 고유 식별번호(Uid) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 df_temp 라는 변수명으로 담으세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다

- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = df['Tenure_Days'].quantile(0.25)
q3 = df['Tenure_Days'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

df_temp = df.drop( df[(df['Tenure_Days'] > upper_fence) | (df['Tenure_Days'] < lower_fence)].index, axis=0)
df_temp = df_temp.drop(columns=['Uid'])
df_temp.reset_index(drop=True, inplace=True)
````
