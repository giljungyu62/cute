### **5. 최근 4주 강수량(mm)(rainfallMm) 컬럼의 이상치는 삭제하지 않고 정상 범위의 경계값으로 대체하고자 합니다.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, clip, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `data`
- 아래 안내대로 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : data
    - IQR 기준(K=3.0)으로 lower_fence, upper_fence 를 산출하세요
    - rainfallMm 값이 두 fence 를 벗어나면 가장 가까운 fence 값으로 바꾸세요(행은 삭제하지 않음)
    - 식별자 컬럼 sampleId 를 삭제한 사본에 적용해 data_temp 에 저장하세요
- **[정답05]** 처리 후 rainfallMm 의 **최댓값**을 소수 둘째 자리까지 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요
- 배점 : 6점

---

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)
q1 = data['rainfallMm'].quantile(0.25)
q3 = data['rainfallMm'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 3.0 * iqr
upper_fence = q3 + 3.0 * iqr

data_temp = data.drop(columns=['sampleId']).copy()
data_temp['rainfallMm'] = data_temp['rainfallMm'].clip(lower=lower_fence, upper=upper_fence)

data_temp['rainfallMm'].max()
````
