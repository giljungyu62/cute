### **5. 그래프로 확인한 결과, 일 평균 걸음 수(steps) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 다음 지침에 따라 이상치를 처리하십시오.**

- 앞 문항 결과 사용 : 2번 `data`
- 대상 데이터프레임 : data
- 일 평균 걸음 수(steps)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=1.5)
- AI모델링에 불필요한 일련번호(serialNo) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 data_temp 라는 변수명으로 보관하세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요 (글자는 따옴표로 감싸기)
- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = data['steps'].quantile(0.25)
q3 = data['steps'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

data_temp = data.drop( data[(data['steps'] > upper_fence) | (data['steps'] < lower_fence)].index, axis=0)
data_temp = data_temp.drop(columns=['serialNo'])
data_temp.reset_index(drop=True, inplace=True)
````

````python
# (5) 정답을 입력하세요 — `정답05` 에 값만 넣고 실행하세요

정답05 = 'reset_index'
````

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

y_train_cat = to_categorical(y_train)
y_valid_cat = to_categorical(y_valid)

model = Sequential([
    Dense(32, activation='elu', input_dim=X_train.shape[1]),
    Dropout(0.2),
    Dense(16, activation='elu'),
    Dense(3, activation='softmax')
])

estop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
cb_list = [estop]
ckpt = ModelCheckpoint('best_model.keras', monitor='val_loss', save_best_only=True)
cb_list.append(ckpt)

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

history = model.fit(X_train, y_train_cat,
                    epochs=30, batch_size=128,
                    validation_data=(X_valid, y_valid_cat),
                    callbacks=cb_list)
````
